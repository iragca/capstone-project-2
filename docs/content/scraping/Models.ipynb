# Models


These models are used for data validation and centralized data design.

## Tweet

### Attributes

| Field                     | Type                                | Default      | Description                                                                                                           |
| ------------------------- | ----------------------------------- | ------------ | --------------------------------------------------------------------------------------------------------------------- |
| `tweet_id`                | `str`                               | **required** | Unique identifier of the tweet.                                                                                       |
| `text`                    | `str`                               | **required** | Raw tweet text content.                                                                                               |
| `status_link`             | `str`                               | **required** | Direct URL to the tweet.                                                                                              |
| `user_id`                 | `str`                               | **required** | Identifier of the authoring user.                                                                                     |
| `is_extremist`            | `Optional[bool]`                    | `False`      | Whether the tweet is flagged as extremist content.                                                                    |
| `is_annotated`            | `Optional[bool]`                    | `False`      | Whether the tweet has been annotated.                                                                                 |
| `in_reply_to_status_link` | `Optional[str]`                     | `None`       | URL of the tweet this one replies to (if any).                                                                        |
| `in_reply_to_status_id`   | `Optional[str]`                     | `None`       | ID of the tweet this one replies to (if any).                                                                         |
| `bookmark_count`          | `int`                               | **required** | Number of bookmarks.                                                                                                  |
| `views`                   | `Optional[int]`                     | `None`       | Number of views (impressions).                                                                                        |
| `retweet_count`           | `int`                               | **required** | Number of retweets.                                                                                                   |
| `favorite_count`          | `int`                               | **required** | Number of likes.                                                                                                      |
| `reply_count`             | `int`                               | **required** | Number of replies.                                                                                                    |
| `quote_count`             | `int`                               | **required** | Number of quote tweets.                                                                                               |
| `conversation_id`         | `str`                               | **required** | ID of the root conversation this tweet belongs to.                                                                    |
| `retweet_status_id`       | `Optional[str]`                     | `None`       | ID of the retweeted tweet if this is a retweet.                                                                       |
| `quoted_status_id`        | `Optional[str]`                     | `None`       | ID of the quoted tweet if this is a quote tweet.                                                                      |
| `community_note`          | `Optional[str]`                     | `None`       | Any community note attached to this tweet.                                                                            |
| `language`                | `str`                               | **required** | ISO language code of the tweet text.                                                                                  |
| `source`                  | `Optional[str]`                     | `None`       | Client/application used to post the tweet.                                                                            |
| `creation_date`           | `str`                               | **required** | Timestamp when the tweet was created (ISO format recommended).                                                        |
| `has_blm_hashtag`         | `bool`                              | **required** | Whether the tweet contains a BLM-related hashtag.                                                                     |
| `fetched_replies`         | `Optional[bool]`                    | `False`      | Whether replies to this tweet have been fetched.                                                                      |
| `is_reply_to_blm`         | `Optional[bool]`                    | `None`       | Whether this is a reply to a BLM-related tweet.                                                                       |
| `is_hateful`              | `Optional[Literal["0","1","2",""]]` | `None`       | Hate speech classification label (`"0" = not hateful`, `"1" = mildly hateful`, `"2" = highly hateful`, `"" = unset`). |

Example:

```python
from src.models import Tweet

tweet = Tweet(
    tweet_id="98765",
    text="Change is coming. #BLM",
    status_link="https://twitter.com/user/status/98765",
    user_id="12345",
    bookmark_count=25,
    retweet_count=10,
    favorite_count=55,
    reply_count=5,
    quote_count=2,
    conversation_id="98765",
    language="en",
    creation_date="2023-07-21T15:30:00Z",
    has_blm_hashtag=True,
    is_hateful="0"
)

print(tweet.text)   # "Change is coming. #BLM"
print(tweet.is_extremist)  # False (default)
```

## User

### Attributes
| Field              | Type                                             | Default         | Description                                            |
| ------------------ | ------------------------------------------------ | --------------- | ------------------------------------------------------ |
| `user_id`          | `str`                                            | **required**    | Unique identifier of the user.                         |
| `username`         | `Optional[str]`                                  | `None`          | Username/handle of the user.                           |
| `name`             | `Optional[str]`                                  | `None`          | Display name of the user.                              |
| `follower_count`   | `Optional[int]`                                  | `0`             | Number of followers.                                   |
| `following_count`  | `Optional[int]`                                  | `0`             | Number of users this account follows.                  |
| `favourites_count` | `Optional[int]`                                  | `0`             | Number of likes/favorites.                             |
| `listed_count`     | `Optional[int]`                                  | `0`             | Number of lists this user is a member of.              |
| `number_of_tweets` | `Optional[int]`                                  | `0`             | Number of tweets the user has posted.                  |
| `is_private`       | `Optional[bool]`                                 | `None`          | Whether the account is private.                        |
| `is_verified`      | `Optional[bool]`                                 | `None`          | Whether the account has a standard verification badge. |
| `is_blue_verified` | `Optional[bool]`                                 | `None`          | Whether the account has a Twitter Blue verification.   |
| `bot`              | `Optional[bool]`                                 | `None`          | Whether the account is identified as a bot.            |
| `location`         | `Optional[str]`                                  | `None`          | User-provided location string.                         |
| `description`      | `str`                                            | **required**    | Profile description (bio).                             |
| `status`           | `Literal["fetched","not fetched","fetching",""]` | `"not fetched"` | Fetch status of this user’s profile.                   |
| `creation_date`    | `Optional[str]`                                  | `None`          | Date the account was created (ISO format preferred).   |
| `friends`          | `Optional[int]`                                  | `0`             | Number of “friends” (usually equivalent to following). |


```python
from src.models import User

user = User(
    user_id="12345",
    username="johndoe",
    name="John Doe",
    follower_count=120,
    following_count=80,
    description="Data scientist and open source enthusiast.",
    status="fetched",
    creation_date="2021-05-12"
)

print(user.username)   # johndoe
print(user.is_verified)  # None (not provided)
```
